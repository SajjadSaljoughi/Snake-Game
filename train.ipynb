```python
import pandas as pd
from sklearn.model_selection import train_test_split
import tensorflow as tf
```

```python
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('/content/drive/MyDrive/Machine Learning/Snake/dataset.csv', header=None)

X = data.iloc[1:, :-1].values.astype("float32")
Y = data.iloc[1:, -1].values

#r = 0,d = 1,l = 2,u = 3
Y[Y == 'r'] = 0
Y[Y == 'd'] = 1
Y[Y == 'l'] = 2
Y[Y == 'u'] = 3

Y = Y.astype("float32")
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2)
```

```python
model = tf.keras.models.Sequential([
    tf.keras.layers.Dense(32, activation=tf.nn.relu),
    tf.keras.layers.Dense(16, activation=tf.nn.relu),
    tf.keras.layers.Dense(4, activation=tf.nn.softmax)
])
```

```python
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
```

```python
history = model.fit(X_train, Y_train,epochs=300)
```

```python
result = model.evaluate(X_test, Y_test)
```

```python
model.save("/content/drive/MyDrive/Machine Learning/Snake/sajjad_saljoughi_snake_ml.h5")
```

```python
plt.plot(history.history["accuracy"], label="Train Accuracy")

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Model Accuracy")

plt.legend()
plt.show()
```

```python
plt.plot(history.history["loss"], label="Train Loss")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Model Loss")

plt.legend()
plt.show()
```
